```python
# Assignment 3
# Problem statement : checking whether a triangle is right angled triangle or not

def is_right_triangle():
    # Taking inputs from the user
    a = float(input("Enter length of first side: "))
    b = float(input("Enter length of second side: "))
    c = float(input("Enter length of third side: "))

    # Store them in a list and sort in ascending order
    sides = [a, b, c]
    sides.sort()

    # Using Pythagoras theorem
    if sides[0] ** 2 + sides[1] ** 2 == sides[2] ** 2:
        print("The triangle is a right-angled triangle.")
    else:
        print("The triangle is not a right-angled triangle.")


# Call the function
is_right_triangle()
```
