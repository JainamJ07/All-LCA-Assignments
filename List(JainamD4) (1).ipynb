```python
numbers = [10,20,30,40,50]
print("Original list:" , numbers)

numbers.append(60)
print("After Appending:", numbers)

numbers.reverse()
print("Reversed List:", numbers)
```

```python
#count the occurnces 

numbers = [10,20,10,30,10,40,50]
elements = int(input("Enter the element to search:"))
count = numbers.count(elements)
print("Number of occurrences:", count)
```
